```python
import numpy as np
arr=np.array([1,2,3,4,5])
print(arr)
print("The dimension of the array:",arr.ndim)
```

```python
arr2=np.array([[1,2,3],[4,5,6]])
print(arr2)
print("The dimension of the array:",arr2.ndim)
print("Shape of the array:",arr2.shape)
```

```python
np.zeros((3,4))
# defining a default array of zeros by specifying the shape of the array
```

```python
np.ones((3,3))
#defining a default array of ones by specifying the shape of the array
```

```python
np.full((3,3),5) 
#defining a default array of the specified value with specifying the shape of the array
```

```python
np.eye(3,3)
#Identity matrix with mentioned shape of the array
```

```python
np.arange(1,10,2) 
# returns an array with starting and ending limit with steps in between
```

```python
np.linspace(0,1,10)
# returns an array with values within 0 and 1 with evenly space amongthe values
```
