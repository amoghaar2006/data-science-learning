```python
import numpy as np
arr=np.array([3,5,6,2,3,4])
mean=np.mean(arr)
print(f"The average of the elements present in the array is:{mean:.2f}")
```

```python
print("The smallest element in the array is:",np.min(arr))
print("The largest element in the array is:",np.max(arr))
```

```python
dev=np.std(arr)
print(f"The standard deviation of the array is:{dev:.2f}")
```

```python
var=np.var(arr)
print(f"The variance of the elements present in the array is:{var:.2f}")
```

```python
med=np.median(arr)
print(f"The median of the element present in the array is:{med:.2f}")
```

```python
print("The sum of the all elements present in the array is:",np.sum(arr))
```

```python
print("The product of the all elements present in the array is:",np.prod(arr))
```

```python
# to compute the percentile of the elements in the array
print("The 1st percentile of the elements present in the array is:",np.percentile(arr,25))
```

```python
print("The index of the largest element in the array is:",np.argmax(arr))
```

```python
print("The index of the smallest element in the array is:",np.argmin(arr))
```

```python
arr2=np.array([1,1,1,2,2,2,5,6,7,8,9])
print("The unique elements present in the array:",np.unique(arr2))
```

```python
log=np.log(arr)
print("The natural log of the array is:",log)
```

```python
print("The cumulative sum of the array:")
np.cumsum(arr)
```

```python
print("the expondential of an array:",np.exp(arr))
```
