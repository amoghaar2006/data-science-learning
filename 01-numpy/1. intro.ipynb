```python
import numpy as np
num=np.array([1,2,3,4,5])
print(num)
```

```python
num2=np.array([[1,2,3],[4,5,6]])
print(num2)
print(num2.shape) #returns the shape of the array. how manu rows and columns
print(num2.ndim)  #returns the dimension of the array
print(type(num2)) #returns the type of the array
```

```python
import sys
print("Numpy array size:",num2.nbytes,"bytes")
print("Python list size:",sys.getsizeof(num2)*len(num2),"bytes")
```
