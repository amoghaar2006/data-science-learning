```python
# numpy arrays are homogenous 
import numpy as np
arr=np.array([1,2,3,4,5])
print(arr.dtype)
```

```python
arr2=arr.astype(np.float64)
print(arr2.dtype)
print(arr2)
```

```python
arr3=arr.astype(np.int32)
print(arr3)
print(arr3.dtype)
```

```python
#smaller data types use less memory
#choosing the appropriate data type ensures that yoy don't lose precision
print(arr2.nbytes)
print(arr3.nbytes)
```

```python
a=np.array([1])
a1=a.astype(np.float32)
print(a1.nbytes)

b=a.astype(np.float64)
print(b.nbytes)
```

```python
arr=np.array([1+2j, 3+3j], dtype="complex128")
print(arr)
```

```python
#string in numpy
arr=np.array(["apple", "banana"], dtype="U10") #Unicode string
print(arr)
#numpy is less effecient that using lists or Python's built in string data types
```

```python
#numpy for storing dictionery
arr = np.array([{'a': 1}, [1, 2, 3], 'hello'], dtype=object)
print(arr)

# This type sacrifices the performance so it should be used only when it is
#absolutely necessary
```
