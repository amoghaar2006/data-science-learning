```python
import pandas as pd
df=pd.read_csv("data_cleaning_sample.csv")
display(df)
```

```python
df2=df.drop_duplicates()
display(df2)
```

```python
df2["Age Group"]=df2["Age"].apply(lambda x: "Adult" if x>=30 else "Minor")
display(df2)
```

```python
print(df2["Gender"].unique())
```

```python
gender_map={"M":"Male","F":"Female"}
df2["Gender"]=df2["Gender"].map(gender_map)
display(df2)
```

```python
df2["City"]=df2["City"].replace({"Delhi":"New Delhi","Mumbai":"Navi Mumbai"})
display(df2)
```
