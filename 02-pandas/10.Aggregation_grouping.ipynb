```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Salary": [85, 90, 78, 85, 92, 88, 75, 80],
    "Age": [23, 25, 30, 22, 28, 26, 21, 27],
    "JoinDate": pd.to_datetime([
        "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
        "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})  

display(df)
```

```python
print("Average Salary of the Employees of each department:")
display(df.groupby("Department")["Salary"].mean())

print("Maximum Salary of an Employee of each department:")
display(df.groupby("Department")["Salary"].max())

print("Minimum salary of an employee of each department:")
display(df.groupby("Department")["Salary"].min())
```

```python
print("The sum of the salary of the employees in each department is:")
print(df.groupby("Department")["Salary"].sum())

sum=df.groupby("Department")["Salary"].sum()

a=sum.index
b=sum.values

plt.bar(a,b, color="blue",label="Sum of salary")
plt.title("Sum of Salaries per Department")
plt.xlabel("Department")
plt.ylabel("Sum of Salary")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
print(df.groupby("Department")["Salary"].agg({"mean","max","min"}))
type(df.groupby("Department")["Salary"].agg({"mean","max","min"}))
```
