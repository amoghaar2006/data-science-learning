```python
import pandas as pd
df=pd.read_csv("data_cleaning_sample.csv")
print(df)
```

```python
print(df.columns)
```

```python
print("No of columns:",len(df.columns))
print("No of rows:",len(df))
```

```python
print("Shape of the dataset:",df.shape)
```

```python
df.isnull() # True null value, False not a null value
```

```python
print("No of null values in each column of the dataset")
df.isnull().sum()
```

```python
print("Percentage of null values in each column of the dataset")
print((df.isnull().sum()/len(df))*100)
```

```python
df.dropna()       # Drop rows with *any* missing values
df.dropna(axis=0) # both are same
```

```python
df.dropna(axis=1) # Drop columns with missing values
```

```python
df.fillna(0) # Replace NaN with 0
```

```python
ds=df["Age"].fillna(df["Age"].mean())
print(ds) # Replace Null values with mean in the age column
```

```python
df.ffill()       # Forward fill fill the null value with the previous recent non null value
```

```python
df.bfill()      # Backward fill fill the null value with the next future non value
```
