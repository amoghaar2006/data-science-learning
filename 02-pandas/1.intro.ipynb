Pandas 
It is a powerful open source Python library used for data manipulation, cleaning and analysis. It provides two main data structures

Series: A one dimensional
DataFrame: A two dimensional labeled table (like an Excel sheet or SQL table)



```python
import pandas as pd
df=pd.DataFrame([10,2,30,40,50,60,70,80,90,100])
print(df)
print(type(df))
```

```python
ds=pd.DataFrame([10,20,30,40],index=["Amoghaa R","Anusha","Ananya","Aisiri"])
print(ds)

```
