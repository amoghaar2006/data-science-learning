```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y1=[num*2 for num in x]
y2=[num**2 for num in x]

plt.subplot(1,2,1)
plt.plot(x,y1)
plt.xlabel("x")
plt.ylabel("Double values")
plt.title("Doubles of x")

plt.subplot(1,2,2)
plt.plot(x,y2)
plt.xlabel("x")
plt.ylabel("Sqaure values")
plt.title("Sqaures of x")


# plt.subplot(2,2,3)
# plt.plot(x,y2)
# plt.title("Sqaures")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y1=[num*2 for num in x]
y2=[num**2 for num in x]
y3=[num*10 for num in x]
y4=[10-num for num in x]
plt.subplot(2,2,1)
plt.plot(x,y1)
plt.xlabel("x")
plt.grid(True)
plt.ylabel("Double values")
plt.title("Doubles of x")

plt.subplot(2,2,2)
plt.plot(x,y2)
plt.grid(True)
plt.xlabel("x")
plt.ylabel("Sqaure values")
plt.title("Sqaures of x")


plt.subplot(2,2,3)
plt.plot(x,y3)
plt.grid(True)
plt.title("Multiples of 10")
plt.xlabel("x")
plt.ylabel("10 Multiples")

plt.subplot(2,2,4)
plt.plot(x,y4)

plt.title("10 minus values of x")
plt.xlabel("x")
plt.grid(True)
plt.ylabel("10 minus x")

plt.tight_layout()
plt.show()
```

```python
fig, axs=plt.subplots(1,2,figsize=(10,4))
axs[0].plot(x,y1)
axs[0].set_title("x*2")

axs[1].plot(x,y2)
axs[1].set_title("x**2")

fig.suptitle("My Comparison Plots")
fig.tight_layout()
plt.show()
```
