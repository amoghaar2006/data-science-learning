```python
import matplotlib.pyplot as plt


years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.bar(years,runs,width=0.5,color="red",label="Runs Scored")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.xticks(range(1990,2011,2))
plt.yticks(range(0,1751,100))
plt.grid(True)
plt.legend()
plt.title("Runs Scored by Sachin Tendulkar based on different years")
plt.tight_layout()
plt.show()
```

```python
import numpy as np

sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
x=np.arange(len(years))
width=0.25
plt.bar(x-width,sachin, width=width, label="Sachin Tendulkar")
plt.bar(x,sehwag, label="Virendar Sehwag", width=width)
plt.bar(x+width, kohli, label="Virat Kohli",width=width)
plt.xticks(x,years) #display the actual years 
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Runs Comparison")
plt.tight_layout()
plt.legend()
plt.show()
```

```python
players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [500+700+1100+1500+1800, 0+200+900+1400+1600, 0+0+500+800+1100, 300+600+800+1100+900]

plt.barh(players,runs_5yrs, color="orange",label="Runs Scored in 5 years")
plt.xlabel("Runs")
plt.ylabel("Players")
plt.title("Runs Scored by Players in 5 years")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
students = ["Asha", "Rahul", "Priya", "Kiran", "Sneha"]
marks = [78, 92, 65, 85, 88]
plt.bar(students,marks, color="red", width=0.5 , label="Marks")
plt.title("Marks of the Students")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.legend()
plt.grid(True, axis="y")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12000, 15000, 11000, 18000, 22000, 19500]
color=["red","green","yellow","orange","blue","pink"]
bars=plt.bar(months, sales, color=color, label="Sales", width=0.7)
plt.title("Monthly Sales Report")
plt.xlabel("Months")
plt.ylabel("Sales (₹)")

plt.bar_label(bars) #helps to to show the number above the bars
plt.grid(True, axis="y")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
technologies = ["Python", "SQL", "Power BI", "Excel", "Tableau"]
beginners = [35, 25, 15, 30, 10]
intermediate = [25, 30, 20, 15, 18]

x=np.arange(len(technologies))
width=0.35
plt.bar(x-width/2, beginners, width=width, label="Beginnners", color="red")
plt.bar(x+width/2, intermediate, width=width, label="Intermediate", color="blue")
plt.xlabel("Technologies")
plt.xticks(x,technologies)
plt.ylabel("Students")
plt.title("Technology Preferences Among Students")
plt.legend()
plt.grid(True, axis="y")
plt.tight_layout()
plt.show()
```
