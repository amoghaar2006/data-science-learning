```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.plot(years,runs, color="blue", marker="o", label="Runs")
plt.title("Sachin Tendulkar's Runs by Years")
plt.xlabel("Years")
plt.ylabel("Runs")
# plt.grid(color="black", linewidth=1, linestyle=":")
plt.xticks(range(1990,2011,2), rotation=45)
# plt.grid(True)
plt.legend()
# plt.tight_layout()
plt.show()
```

```python
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years,kohli, label="Virat Kohli",color="red",marker="o",linestyle="--")
plt.plot(years,sehwag,label="Virender Sehwag", color="blue",marker="o",linestyle="-.")

plt.title("Runs Comparision by Years")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.xticks(range(1990,2011,2),rotation=45)
plt.yticks(range(0,2101,300))
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]

kohli = [1200, 1400, 1800, 1600, 1700, 800, 1100, 1300, 1900, 1500]

rohit = [900, 1100, 1300, 1500, 1400, 700, 1000, 1600, 1800, 1700]

sehwag = [1500, 1300, 1100, 900, 700, 500, 300, 0, 0, 0]

plt.plot(years,kohli, 'ro--',label="Virat Kohli")
plt.plot(years,rohit, 'b^--',label="Rohit Sharma")
plt.plot(years,sehwag,'y*-', label="Virender Sehwag" )

plt.title("Runs Comparison")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
departments = ["HR", "IT", "Sales", "Finance", "Marketing"]
employees = [12, 35, 24, 18, 20]

plt.plot(departments, employees, color="blue", marker="o")
plt.title("Employee Count by Department")
plt.ylabel("Number of Employees")
plt.xlabel("Departments")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [28, 30, 29, 32, 31, 34, 33]
plt.plot(days, temperature, color="red", marker="o")
plt.title("Weekly Temperature Variation")
plt.ylabel("Temperature (°C)")
plt.xlabel("Days")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

website_visitors = [1200, 1500, 1350, 1800, 2200, 2500,
                    2300, 2800, 3100, 2900, 3500, 4000]

app_visitors = [900, 1100, 1250, 1400, 1700, 1900,
                2100, 2000, 2400, 2700, 3000, 3300]
plt.title("Monthly Visitor Comparison")
plt.xlabel("Months")
plt.ylabel("Number of Visitors")
plt.plot(months, website_visitors, "ro--", label="Website Visitors")
plt.plot(months, app_visitors, "b*-", label="App Visitors")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
years = [2021, 2022, 2023, 2024, 2025]

company_A = [250000, 300000, 280000, 380000, 450000]
company_B = [200000, 260000, 320000, 350000, 400000]
company_C = [180000, 220000, 250000, 310000, 370000]
plt.title("Annual Revenue Comparison")
plt.xlabel("Year")
plt.ylabel("Revenue (₹)")
plt.plot(years, company_A, color="blue", marker="o", label="Company A")
plt.plot(years, company_B, color="red", marker="*", label="Company A")
plt.plot(years, company_C, color="green", marker="^", label="Company A")
plt.legend()
plt.yticks(range(100000,500000, 50000))
plt.xticks(years,years)
plt.grid(True)
plt.tight_layout()
plt.show()
```
