```python
import matplotlib.pyplot as plt

labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]
plt.title("Pie Chart representing the runs scored by batters")
colors=["pink","red","green","yellow"]
plt.pie(runs,labels=labels,colors=colors, wedgeprops={"edgecolor":"black","linestyle":"solid","linewidth":2})
plt.tight_layout()
plt.show()
```

```python
explode = [0.1, 0, 0, 0]  # Only highlight Sachin's slice
 
plt.pie(runs, labels=labels, explode=explode, colors=colors)
plt.title("Exploded Pie Example")
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]
plt.title("Pie Chart representing the runs scored by batters")
colors=["pink","red","green","yellow"]
plt.pie(runs,labels=labels,colors=colors, shadow=True, startangle=150, autopct="%1.1f%%", wedgeprops={"linestyle":"solid","edgecolor":"black","linewidth":3}) # to rotate the pie chart to 150 degrees
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
activities = ["Studying", "Sleeping", "Playing", "Watching TV", "Other"]
hours = [5, 7, 3, 2, 7]

plt.pie(hours, labels=activities, colors=["pink", "red", "green","yellow","blue"], autopct="%1.1f%%", wedgeprops={"edgecolor":"black"})
plt.title("Daily Time Allocation")
# plt.axis("equal")
plt.tight_layout()
plt.show()

```

```python
companies = ["Company A", "Company B", "Company C", "Company D", "Others"]
market_share = [35, 25, 18, 12, 10]

plt.pie(market_share, labels=companies, colors=["red","green","blue","yellow","orange"], shadow=True, autopct="%1.1f%%")
plt.title("Market Share of Companies")
plt.axis("equal")
plt.tight_layout()
plt.show()
```

```python
explode=[0,0,0,0,0,0.2]
expenses = ["Food", "Transport", "Rent", "Education", "Shopping", "Savings"]
amounts = [4500, 1500, 6000, 2500, 2000, 3500]

plt.title("Monthly Budget Distribution")
plt.pie(amounts, labels=expenses, colors=["red","green","blue","yellow","orange","purple"],  explode=explode, autopct="%1.1f%%", shadow=True, wedgeprops={"linewidth":2,"edgecolor":"black"})
plt.axis("equal")
plt.tight_layout()
plt.show()

```
