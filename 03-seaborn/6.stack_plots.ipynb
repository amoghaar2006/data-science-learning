```python
import matplotlib.pyplot as plt
activites=["Studying","Playing","Watching TV","Sleeping"]
timespent=[3,2,2,5]
plt.figure(figsize=(6,6))
plt.pie(timespent,labels=activites, colors=["red","green","blue","yellow"], shadow=True, autopct="%1.1f%%", wedgeprops={"linewidth":3,"edgecolor":"black","linestyle":"solid"})
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days=[1,2,3,4,5,6,7]
studying=[3,4,3,3,4,3,4]
playing=[2,2,1,1,2,3,2]
watching_tv=[2,1,2,1,1,1,2]
sleeping=[5,5,6,5,6,5,5]

labels=["studying","playing","watching_tv","sleeping"]
colors=["red","green","blue","yellow"]
plt.figure(figsize=(6,6))
plt.stackplot(days,studying,playing,watching_tv,sleeping,labels=labels,colors=colors,alpha=0.4)
plt.legend(loc="upper left")
plt.xlabel("Days of the week")
plt.yticks(range(1,13,1))
plt.ylabel("Hours")
plt.grid(True)
plt.title("Hours spend on different activities")
plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5, 6, 7]

studying = [3, 4, 3, 5, 4, 3, 4]
playing = [2, 1, 2, 1, 2, 3, 2]
sleeping = [7, 6, 7, 6, 7, 8, 7]
other = [12, 13, 12, 12, 11, 10, 11]
labels=["studying","playing","sleeping","other"]
colors=["red","green","blue","yellow"]
plt.title("Daily Time Allocation")
plt.xlabel("Days")
plt.ylabel("Hours")

plt.stackplot(days,studying,playing,sleeping,other, colors=colors, labels=labels)
plt.grid(True, axis="y")
plt.legend(loc="upper right")
plt.tight_layout()
plt.show()


```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

organic = [120, 150, 180, 200, 240, 280]
social = [80, 90, 100, 120, 140, 160]
direct = [60, 70, 65, 90, 100, 110]
referral = [30, 40, 45, 50, 60, 75]
labels=["organic","social","direct","referral"]

colors=["red","green","blue","yellow","purple"]
plt.title("Monthly Website Traffic")
plt.xlabel("Months")
plt.ylabel("Number of Visitors")
plt.stackplot(months, organic, social, direct, referral, labels=labels, colors=colors)
plt.legend()
plt.grid(True, axis="y")
plt.tight_layout()
plt.show()
```
