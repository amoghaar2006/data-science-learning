```python
import seaborn as sns
import matplotlib.pyplot as plt
print(sns.get_dataset_names())
```

```python
data=sns.load_dataset("tips")
data.head()
```

```python
data.tail()
```

```python
print("No of columns:",len(data.columns))
print("No of rows:",len(data))
```

```python
sns.lineplot(x="total_bill",y="tip",data=data)
plt.grid(True)
plt.title("Total Bill v/s Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip offered")
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(x="total_bill", y="tip", data=data, hue="time")
plt.title('Scatter Plot with Color by Time')
plt.show()
```

```python
sns.barplot(x="day", y="total_bill", data=data)
plt.title('Total Bill per Day')
plt.show()
```

```python
sns.boxplot(x="day", y="total_bill", data=data)
plt.title('Boxplot of Total Bill per Day')
plt.show()
```

```python
flight=sns.load_dataset("flights")
pivot_table=flight.pivot(index="month", columns="year", values="passengers")

sns.heatmap(pivot_table, annot=True, fmt="d", cmap="YlGn")
plt.title("HeatMap of Pasengers")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
info={
    "age": [22, 25, 47, 52, 46, 56, 55, 60, 34, 43],
    "salary": [25000, 27000, 52000, 60000, 58000, 62000, 61000, 65000, 38000, 45000],
    "gender": ["M", "F", "M", "F", "F", "M", "M", "F", "F", "M"]
}
df = pd.DataFrame(info)

sns.scatterplot(x="age", y="salary", hue="gender", data=df)
plt.title("Salary vs Scatter Plot")
plt.tight_layout()
plt.grid(True)
plt.show()
```
